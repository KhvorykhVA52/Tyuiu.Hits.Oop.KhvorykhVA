```
using System;
using System.Collections.Generic;

public abstract class Component
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Component(string name, decimal price)
    {
        Name = name;
        Price = price;
    }

    public virtual string GetInfo()
    {
        return $"{Name} - {Price} руб.";
    }

    public string GetInfo(bool detailed)
    {
        if (detailed)
        {
            return $"{Name} Цена: {Price} руб. Детальная информация";
        }
        return GetInfo();
    }

    public string GetInfo(string category)
    {
        return $"{Name} Цена: {Price} руб. Категория: {category}";
    }
}

public interface IInstallable
{
    void Install();
}

public class CPU : Component, IInstallable
{
    public int Cores { get; set; }

    public CPU(string name, decimal price, int cores) : base(name, price)
    {
        Cores = cores;
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()} Ядра: {Cores}";
    }

    public void Install() { }
}

public class RAM : Component, IInstallable
{
    public int Capacity { get; set; }

    public RAM(string name, decimal price, int capacity) : base(name, price)
    {
        Capacity = capacity;
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()} Объем: {Capacity}GB";
    }

    public void Install() { }
}

public class GPU : Component, IInstallable
{
    public int Memory { get; set; }

    public GPU(string name, decimal price, int memory) : base(name, price)
    {
        Memory = memory;
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()} Память: {Memory}GB";
    }

    public void Install() { }
}

public class HDD : Component, IInstallable
{
    public int Capacity { get; set; }

    public HDD(string name, decimal price, int capacity) : base(name, price)
    {
        Capacity = capacity;
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()} Объем: {Capacity}GB";
    }

    public void Install() { }
}

public class Computer
{
    private List<Component> components = new List<Component>();

    public void AddComponent(Component component)
    {
        components.Add(component);
    }

    public void RemoveComponent<T>(T component) where T : Component
    {
        components.Remove(component);
        Console.WriteLine($"Удален компонент: {component.Name}");
    }

    public decimal GetTotalPrice()
    {
        decimal total = 0;
        for (int i = 0; i < components.Count; i++)
        {
            total = total + components[i].Price;
        }
        return total;
    }

    public void GetTotalPrice(bool showMessage)
    {
        decimal total = GetTotalPrice();
        if (showMessage)
        {
            Console.WriteLine($"Общая стоимость: {total} руб.");
        }
    }

    public void ShowComponents()
    {
        foreach (var component in components)
        {
            Console.WriteLine(component.GetInfo());
        }
    }

    public void ShowAllComponents(bool detailed)
    {
        foreach (var component in components)
        {
            Console.WriteLine(component.GetInfo(detailed));
        }
    }
}

public static class ComputerFactory
{
    public static Computer CreateGamingPC()
    {
        var pc = new Computer();
        pc.AddComponent(new CPU("Intel i7", 25000, 8));
        pc.AddComponent(new RAM("DDR4", 8000, 16));
        pc.AddComponent(new GPU("RTX 4080", 60000, 16));
        return pc;
    }

    public static Computer CreateOfficePC()
    {
        var pc = new Computer();
        pc.AddComponent(new CPU("Intel i3", 15000, 4));
        pc.AddComponent(new RAM("DDR4", 4000, 8));
        pc.AddComponent(new HDD("Seagate", 5000, 2000));
        return pc;
    }
}

{
    {
        var gaming = ComputerFactory.CreateGamingPC();
        Console.WriteLine("Gaming PC:");
        gaming.ShowComponents();
        Console.WriteLine($"Total: {gaming.GetTotalPrice()}");

        var office = ComputerFactory.CreateOfficePC();
        Console.WriteLine("\nOffice PC:");
        office.ShowComponents();
        Console.WriteLine($"Total: {office.GetTotalPrice()}");
    }
}
```
