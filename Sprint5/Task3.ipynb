```
using System;
using System.Collections.Generic;
using System.Linq;

public delegate void AccountAction(string message);

public interface ITransferable
{
    void Transfer(BankAccount target, decimal amount);
}

public interface ILoanable
{
    void TakeLoan(decimal amount);
    void PayLoan(decimal amount);
}

public class BankAccount
{
    public string AccountNumber { get; set; }
    public decimal Balance { get; set; }
    public string OwnerName { get; set; }
    public DateTime CreatedDate { get; set; }
    public string Currency { get; set; }
    public bool IsActive { get; set; }
    
    public decimal InterestRate { get; set; }
    public int TransactionCount { get; set; }
    public string BankName { get; set; }
    public List<string> TransactionHistory { get; set; }
    
    public event AccountAction OnTransaction;
    
    public BankAccount(string accountNumber, decimal balance, string ownerName)
    {
        AccountNumber = accountNumber;
        Balance = balance;
        OwnerName = ownerName;
        CreatedDate = DateTime.Now;
        Currency = "RUB";
        IsActive = true;
        InterestRate = 5.0m;
        TransactionCount = 0;
        BankName = "Центральный Банк";
        TransactionHistory = new List<string>();
    }
    
    public virtual void Deposit(decimal amount)
    {
        Balance += amount;
        TransactionCount++;
        TransactionHistory.Add($"Внесение: +{amount}");
        OnTransactionHandler($"Внесено {amount} на счет {AccountNumber}");
    }
    
    public virtual void Withdraw(decimal amount)
    {
        if (amount > Balance)
            throw new Exception("Недостаточно средств");
        Balance -= amount;
        TransactionCount++;
        TransactionHistory.Add($"Снятие: -{amount}");
        OnTransactionHandler($"Снято {amount} со счета {AccountNumber}");
    }
    
    public void ApplyInterest()
    {
        decimal interest = Balance * InterestRate / 100;
        Balance += interest;
        OnTransactionHandler($"Начислены проценты: {interest}");
    }
    
    public void ShowTransactionHistory()
    {
        Console.WriteLine($"История операций счета {AccountNumber}:");
        TransactionHistory.ForEach(t => Console.WriteLine($"  {t}"));
    }
    
    public void ChangeBank(string newBank)
    {
        BankName = newBank;
        OnTransactionHandler($"Банк изменен на: {newBank}");
    }
    
    public void ShowBalance()
    {
        Console.WriteLine($"Баланс счета {AccountNumber}: {Balance}");
    }
    
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Счет: {AccountNumber}, Владелец: {OwnerName}, Баланс: {Balance}, Банк: {BankName}");
    }
    
    protected void OnTransactionHandler(string message)
    {
        OnTransaction?.Invoke(message);
    }
}

public class StudentAccount : BankAccount, ITransferable
{
    public string University { get; set; }
    public int Year { get; set; }
    public string StudentId { get; set; }
    public decimal MonthlyLimit { get; set; }
    
    public decimal Scholarship { get; set; }
    public bool HasDormitory { get; set; }
    public List<string> Discounts { get; set; }
    
    public StudentAccount(string accountNumber, decimal balance, string ownerName, string university, int year)
        : base(accountNumber, balance, ownerName)
    {
        University = university;
        Year = year;
        StudentId = $"STU{DateTime.Now:yyyyMMddHHmmss}";
        MonthlyLimit = 50000;
        Scholarship = 3000;
        HasDormitory = true;
        Discounts = new List<string> { "Транспорт", "Канцтовары" };
    }
    
    void ITransferable.Transfer(BankAccount target, decimal amount)
    {
        Withdraw(amount);
        target.Deposit(amount);
        Console.WriteLine($"Перевод {amount} со студенческого счета");
    }
    
    public void ReceiveScholarship()
    {
        Balance += Scholarship;
        OnTransactionHandler($"Получена стипендия: {Scholarship}");
    }
    
    public void AddDiscount(string discount)
    {
        Discounts.Add(discount);
        OnTransactionHandler($"Добавлена скидка: {discount}");
    }
    
    public void ShowStudentInfo()
    {
        Console.WriteLine($"Студент: {OwnerName}, Университет: {University}, Стипендия: {Scholarship}");
    }
    
    public void ApplyDiscount(decimal discount)
    {
        Balance += discount;
        Console.WriteLine($"Студенческая скидка {discount} применена");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Университет: {University}, Курс: {Year}, Скидки: {string.Join(", ", Discounts)}");
    }
}

public class BusinessAccount : BankAccount, ILoanable
{
    public string CompanyName { get; set; }
    public string TaxId { get; set; }
    public decimal CreditLimit { get; set; }
    public decimal CurrentLoan { get; set; }
    
    public string BusinessType { get; set; }
    public int EmployeeCount { get; set; }
    public List<string> Partners { get; set; }
    
    public BusinessAccount(string accountNumber, decimal balance, string ownerName, string companyName)
        : base(accountNumber, balance, ownerName)
    {
        CompanyName = companyName;
        TaxId = $"TAX{DateTime.Now:yyyyMMddHHmmss}";
        CreditLimit = 1000000;
        CurrentLoan = 0;
        BusinessType = "ООО";
        EmployeeCount = 10;
        Partners = new List<string>();
    }
    
    public void TakeLoan(decimal amount)
    {
        if (amount > CreditLimit - CurrentLoan)
            throw new Exception("Превышен кредитный лимит");
        
        Balance += amount;
        CurrentLoan += amount;
        OnTransactionHandler($"Взят бизнес-кредит: {amount}");
    }
    
    public void PayLoan(decimal amount)
    {
        if (amount > CurrentLoan)
            throw new Exception("Сумма платежа превышает задолженность");
        
        Balance -= amount;
        CurrentLoan -= amount;
        OnTransactionHandler($"Погашен кредит: {amount}");
    }
    
    public void AddPartner(string partner)
    {
        Partners.Add(partner);
        OnTransactionHandler($"Добавлен партнер: {partner}");
    }
    
    public void CalculateTax()
    {
        decimal tax = Balance * 0.15m;
        Console.WriteLine($"Налог к уплате: {tax}");
    }
    
    public void HireEmployee()
    {
        EmployeeCount++;
        OnTransactionHandler($"Нанят сотрудник. Всего: {EmployeeCount}");
    }
    
    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Компания: {CompanyName}, Сотрудники: {EmployeeCount}, Кредит: {CurrentLoan}/{CreditLimit}");
    }
}

public class AccountService
{
    private readonly BankAccount _account;

    public AccountService(BankAccount account)
    {
        _account = account;
    }

    public void ProcessAccount()
    {
        _account.DisplayInfo();
        _account.ShowBalance();
    }
}

{
    {
        List<BankAccount> accounts = new List<BankAccount>();
        
        StudentAccount student = new StudentAccount("STU001", 1000, "Иван", "МГУ", 2);
        BusinessAccount business = new BusinessAccount("BUS001", 50000, "Петр", "ООО Ромашка");
        
        student.OnTransaction += msg => Console.WriteLine($"🔔 {msg}");
        business.OnTransaction += msg => Console.WriteLine($"🏢 {msg}");
        
        AccountAction notify = message => Console.WriteLine($"📢 Оповещение: {message}");
        student.OnTransaction += notify;
        business.OnTransaction += notify;
        
        accounts.Add(student);
        accounts.Add(business);
        
        Console.WriteLine("=== ОПЕРАЦИИ ===");
        student.Deposit(5000);
        student.ReceiveScholarship();
        student.AddDiscount("Кафе");
        
        business.Deposit(20000);
        business.HireEmployee();
        business.AddPartner("ИП Сидоров");
        
        var richAccounts = accounts.Where(a => a.Balance > 10000);
        Console.WriteLine("\n=== СЧЕТА С БАЛАНСОМ > 10000 ===");
        foreach (var acc in richAccounts)
            acc.DisplayInfo();
        
        ITransferable transferStudent = student;
        transferStudent.Transfer(new BankAccount("ACC002", 500, "Сергей"), 1000);
        
        ILoanable loanBusiness = business;
        loanBusiness.TakeLoan(50000);
        
        Console.WriteLine("\n=== ИНФОРМАЦИЯ ===");
        new AccountService(student).ProcessAccount();
        Console.WriteLine("---");
        new AccountService(business).ProcessAccount();
        
        student.ShowTransactionHistory();
    }
}
```
